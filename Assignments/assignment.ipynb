```python
def dataset_eda(df):

    shape_of_df = df.shape
    col_names = df.columns.tolist()
    col_dtypes = df.dtypes
    missing_val = df.isnull().sum()
    dup_val = df.duplicated().sum()
    uniq_val = df.nunique()
    stat = df.describe()

    return shape_of_df, col_names, col_dtypes, missing_val, dup_val, uniq_val, stat

dataset_eda(df)
```

```python
target = "Loan_Status"
```

```python
df.columns.to_list()
```

```python
def feat_engg(df, num_cols, cat_cols):

    for col in num_cols:
        if df[col].isnull().sum() > 0:
            df[col] = df[col].fillna(df[col].mean())

    for col in cat_cols:
        if df[col].isnull().sum() > 0:
            df[col] = df[col].fillna(df[col].mode()[0])


    for col in num_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = (df[col] < lower) | (df[col] > upper)

        no_of_out = outliers.sum()

        df[col] = df[col].clip(lower, upper)

    return df

cat_cols = df.select_dtypes(include = "object").columns.drop(target).tolist()
num_cols = df.select_dtypes(include = np.number).columns.tolist()
df = feat_engg(df, num_cols, cat_cols)

print(num_cols)
print(cat_cols)

```
