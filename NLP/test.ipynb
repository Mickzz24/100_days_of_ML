```python
import joblib

import re

load_pre = joblib.load("preprocessing.joblib")

bow_vectorizer = joblib.load("bow_vectorizer.joblib")

bow_model = joblib.load("bow_model.joblib")
```

```python
stop_words = load_pre['stopwords']
```

```python
tweetData = "India played an amazing match today."

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r"[^a-zA-Z]", " ", text)

    words = text.split()

    words = [stemmer.stem(word) for word in words if word not in stop_words]


    return " ".join(words)

cleaned_text = clean_text(tweetData)
```

```python
vector = bow_vectorizer.transform([cleaned_text])
```

```python
bow_model.predict(vector)
```
